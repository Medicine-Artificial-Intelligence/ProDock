```python
import prolif as plf
import MDAnalysis as mda
import tempfile
import time
import os
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from IPython.display import display

# Step 1: Load protein and ligand
protein_path = "/liu1/Son/Projects/RHOA/MD/0_1000/peptide_1_1/rhoa_lowest_energy.pdb"
ligand_sdf = "/liu1/Son/Projects/RHOA/MD/0_1000/peptide_1_1/peptide_lowest_energy.sdf"

u = mda.Universe(protein_path)
protein_mol = plf.Molecule.from_mda(u)
pose_iterable = plf.sdf_supplier(ligand_sdf)

# Step 2: Run fingerprint
fp = plf.Fingerprint()
result = fp.run_from_iterable(pose_iterable, protein_mol)

# Step 3: Plot lignetwork (returns IPython.display.HTML)
html_obj = fp.plot_lignetwork(pose_iterable[0])

# Step 4: Write HTML to temp file
html_str = html_obj.data  # Extract raw HTML from IPython HTML object
with tempfile.NamedTemporaryFile(suffix=".html", delete=False, mode="w") as f:
    f.write(html_str)
    html_path = f.name

# Step 5: Screenshot via Selenium
options = Options()
options.headless = True
options.add_argument("--window-size=1000,900")

driver = webdriver.Chrome(options=options)
driver.get("file://" + html_path)
time.sleep(2)  # Wait for JS to render

screenshot_path = "/liu1/Son/Projects/TYK2/result_combined/gnina_centroid/natural_18k_14933.png"
driver.save_screenshot(screenshot_path)
driver.quit()

print(f"[✔] Screenshot saved to: {screenshot_path}")
```
